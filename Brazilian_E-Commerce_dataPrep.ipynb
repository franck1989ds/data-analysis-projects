# Orders

```python
name = "olist_orders_dataset"
filename = path + name + ".csv"

orders = pd.read_csv(filename)
```

```python
orders.head()
```

```python
orders.shape
```

```python
orders.dtypes
```

All the parameters are str, but they do represent a datetime.

```python
def convert_to_datetime(df,column):
    df[column] = pd.to_datetime(df[column], errors = "coerce")
```

```python
columns_to_convert = orders.columns[3:].tolist()
for col in columns_to_convert:
    convert_to_datetime(orders,col)

orders.dtypes
```

```python
# check for nulls and duplicates
orders.isna().sum()
```

```python
cols = [
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
]

orders.groupby("order_status")[cols].agg(lambda x: x.isna().sum())
```

Most missing dates are consistent with the order status, such as shipped orders without a delivery date.

However, delivered orders include 14 missing approval dates, 2 missing carrier handover dates, and 8 missing delivery dates. These might be excluded later in the analysis.

```python
orders.duplicated().sum()
```

```python
orders["order_id"].duplicated().sum()
```

```python
orders["order_status"].value_counts()
```

```python
# incosistences in the dates.
orders.loc[
    orders["order_delivered_customer_date"]
    < orders["order_purchase_timestamp"]
]
```

```python
filename = path_dest + name + "_cleaned.parquet"

orders.to_parquet(filename)
```
